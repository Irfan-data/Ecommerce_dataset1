```python
# here  i am importing the all libraries of the python for cleaning
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
# here i am apploading / calling the dataset

df = pd.read_csv("/content/messy_ecommerce_30000_rows.csv")
#v here i am calling the top five rows
df.head()
```

```python
df.columns  # here i want to see all the columns of the table
```

```python
# Now i want to checks the types
# df.dtypes to check data types of all the columns
df.dtypes
```

```python
# object data type
# object data_types is a general data types. used for storing text , mixed types python objects
```

```python
# here i want to chanege multiple coluns data types togethor
for col in(['Customer_Name','Product','Category','Payment_Method','City']):
  df[col] = df[col].astype('string')
```

```python
df.dtypes
```

```python
for col in(['Price','Total_Sale']):
  df[col]=df[col].astype('int')
```

```python
df.dtypes
```

```python
#here i am filling where is null values nin numerical columns
for col in(['OrderID','Price','Total_Sale','Quantity']):
  df[col]=df[col].interpolate(method='linear')
```

```python
df.head()
```

```python
df['Discount']=pd.to_numeric(df['Discount'],errors='coerce')
```

```python
df.dtypes
```

```python
# here  i am converting the data into proper data formate
df['Order_Date']=pd.to_datetime(df['Order_Date'])
```

```python
df.dtypes # here i am chacking the datatypes of the each values
```

```python
df.head()
```

```python
#here i used .str.strip(function for trim) and capitalize for change case
for col in(['Customer_Name','Product','Category','Payment_Method','City']):
  df[col]=df[col].str.strip().str.capitalize()
```

```python
df.head()
```

```python
# here i am fulling missing values into columns where is null
for col in(['OrderID','Quantity','Price','Total_Sale','Discount']):
  df[col]=df[col].interpolate(method='linear')
```

```python
df.head()
```

```python
df.drop_duplicates(inplace=True)

#here i removed the duplicates values
```

```python
print(df.isnull().sum()) # here i am chacking that is any null values in the table or not
```

```python
df.head()
```

```python
df.to_csv('clean_ecommerce_data_set')

df
```

```python
Total_Sale_by_Date=df.groupby('Order_Date')['Total_Sale'].sum()
print(Total_Sale_by_Date)

plt.figure(figsize=(12,6))
plt.plot(Total_Sale_by_Date.index,Total_Sale_by_Date.values,marker='o',linestyle='-')
plt.title('Total Sale by Date')
plt.xlabel('Date')
plt.ylabel('Total Sale')
plt.grid(True)
plt.show()
```

```python
Total_Sale_by_region=df.groupby('City')['Total_Sale'].sum()
print(Total_Sale_by_region)


plt.figure(figsize=(12,6))
plt.bar(Total_Sale_by_region.index,Total_Sale_by_region.values,color=['red','green','blue'])
plt.title('Total Sale by Region')
plt.xlabel('Region')
plt.ylabel('Total Sale')
plt.show()
```

```python
Total_Sale_by_Category=df.groupby('Category')['Total_Sale'].sum()
print(Total_Sale_by_Category)

plt.figure(figsize=(12,6))
plt.pie(Total_Sale_by_Category.values,labels=Total_Sale_by_Category.index,autopct='%1.1f%%',startangle=90)
plt.title('Total Sale by Category')
plt.show()
```

```python
Total_Sale_by_Payment_Method=df.groupby('Payment_Method')['Total_Sale'].sum()
print(Total_Sale_by_Payment_Method)

plt.figure(figsize=(12,6))
plt.pie(Total_Sale_by_Payment_Method.values,labels=Total_Sale_by_Payment_Method.index,autopct='%1.1f%%',startangle=90)
plt.title('Total Sale by Payment Method')
plt.show()
```

```python
Total_Sale_by_City=df.groupby('City')['Total_Sale'].sum()
print(Total_Sale_by_City)

plt.figure(figsize=(12,6))
plt.barh(Total_Sale_by_City.index,Total_Sale_by_City.values,color=['red','green','blue'])
plt.title('Total Sale by City')
plt.xlabel('Total Sale')
plt.show()
```
